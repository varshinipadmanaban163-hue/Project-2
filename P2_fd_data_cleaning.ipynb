### Handling Missing Values
- Numeric : Impute with Mean/Median
- Category : Impute with Mode
- 0 - 20% : impute with mean/meadian
- 0 - 60% : impute with mise imputer

```python
df.isnull().mean()*100
```

```python
num_columns= ['Customer_Age','Delivery_Time_Min','Distance_km','Order_Value','Discount_Applied', 'Delivery_Rating',"Final_Amount","Restaurant_Rating","Profit_Margin"]
for i in num_columns:
    Q1 = df[i].quantile(0.25)
    Q3 = df[i].quantile(0.75)
    IQR = Q3-Q1
    lower_bound =Q1 -(1.5*IQR)
    upper_bound =Q3 +(1.5*IQR)

    outliers = df[(df[i] < lower_bound) | (df[i] > upper_bound)]
    print(f"  {i} — {len(outliers)} outliers")
```

```python
mode_columns=["Order_ID","Customer_Gender","Customer_ID","Restaurant_ID","Restaurant_Name","Delivery_Partner_ID","City","Area","Cuisine_Type","Payment_Mode","Order_Status","Order_Day","Peak_Hour",'Order_Date', 'Order_Time']
for i in mode_columns:
    df[i] = df[i].fillna(df[i].mode()[0])
```

```python
num_columns= ['Delivery_Time_Min','Distance_km', 'Delivery_Rating',"Restaurant_Rating","Profit_Margin"]
df[num_columns] = df[num_columns].fillna(df[num_columns].median())
```

```python
df['Order_Value'] = df['Order_Value'].fillna(
    df.groupby('Restaurant_ID')['Order_Value'].transform('median'))
```

```python
df['Discount_Applied'] = df['Discount_Applied'].fillna(
    df.groupby('Restaurant_ID')['Discount_Applied'].transform('median'))
```

```python
df['Final_Amount'] = df['Final_Amount'].fillna(
    df['Order_Value'] - df['Discount_Applied']
)
```

```python
df["Cancellation_Reason"]=df["Cancellation_Reason"].fillna("Not Cancelled")
```

```python
df.isnull().mean()*100
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer


df_mice_subset = df[['Customer_Age']]

mice_imputer = IterativeImputer(random_state=42)

df_mice_imputed_array = mice_imputer.fit_transform(df_mice_subset)

df_mice_imputed = pd.DataFrame(
    df_mice_imputed_array,
    columns=["Customer_Age"],
    index=df.index
)

df['Customer_Age'] = df_mice_imputed['Customer_Age'].round().astype(int)
```

```python
df.isnull().mean()*100
```

### Outlier Detection

```python
num_columns= ['Customer_Age','Delivery_Time_Min','Distance_km','Order_Value','Discount_Applied', 'Delivery_Rating',"Final_Amount","Restaurant_Rating","Profit_Margin"]
for i in num_columns:
    Q1 = df[i].quantile(0.25)
    Q3 = df[i].quantile(0.75)
    IQR = Q3-Q1
    lower_bound =Q1 -(1.5*IQR)
    upper_bound =Q3 +(1.5*IQR)

    outliers = df[(df[i] < lower_bound) | (df[i] > upper_bound)]
    print(f"  {i} — {len(outliers)} outliers")
```

```python
Q1 = df["Discount_Applied"].quantile(0.25)
Q3 = df["Discount_Applied"].quantile(0.75)

IQR = Q3 - Q1

lowerbound = Q1 - 1.5 * IQR
upperbound = Q3 + 1.5 * IQR

df["Discount_Applied"] = df["Discount_Applied"].clip(lowerbound, upperbound)
```

```python
df["Discount_Applied"] = df["Discount_Applied"].clip( lower=lowerbound,
    upper=upperbound  )
```
